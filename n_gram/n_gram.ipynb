```python
import nltk
from collections import Counter
from nltk.util import ngrams
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
text = """
Natural language processing is interesting.
Natural language processing is useful.
I love natural language processing.
"""
```

```python
tokens = word_tokenize(text.lower())

tokens = [word for word in tokens if word.isalpha()]

print("\n Original Text:")
print(tokens)

print("\n Tokens:")
print(tokens)
```

```python
unigrams = list(ngrams(tokens, 1))
unigram_freq = Counter(unigrams)

print("\n Unigrams and Frequencies:")

for gram, count in unigram_freq.items():
    print(gram, ":", count)
```

```python
bigrams = list(ngrams(tokens, 2))
bigram_freq = Counter(bigrams)

print("\n Bigrams and Frequencies:")

for gram, count in bigram_freq.items():
    print(gram, ":", count)
```

```python
trigrams = list(ngrams(tokens, 3))
trigram_freq = Counter(trigrams)

print("\n Trigrams and Frequencies:")

for gram, count in trigram_freq.items():
    print(gram, ":", count)
```

```python
#---Calculate a bigram conditional probability
#    P("language" | "natural") ---#

previous_word = "natural"
current_word = "language"

bigram_count = bigram_freq[(previous_word, current_word)]

previous_word_count = unigram_freq[(previous_word,)]

if previous_word_count > 0:
    probability = bigram_count / previous_word_count
else:
    probability = 0.0

print("\nBigram Probability:")
print(
    f"P({current_word} | {previous_word}) = "
    f"{bigram_count}/{previous_word_count} = {probability:.4f}"
)
```
