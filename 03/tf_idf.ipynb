```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
documents = [
    "I love natural language processing",
    "I love machine learning",
    "Natural language processing is interesting"
]
```

```python
vectorizer = TfidfVectorizer()
```

```python
tfidf_matrix = vectorizer.fit_transform(documents)
```

```python
features = vectorizer.get_feature_names_out()
```

```python
print("\n Vocabulary:")
print(features)
```

```python
print("\n TF-IDF Feature Matrix:")
print(tfidf_matrix.toarray())
```

```python
print("\nDocument-Term TF-IDF Matrix:")

print(f"{'Document':<12}", end="")
for feature in features:
    print(f"{feature:<14}", end="")
print()

for i, row in enumerate(tfidf_matrix.toarray()):
    print(f"{'Document ' + str(i + 1):<12}", end="")
    for value in row:
        print(f"{value:<14.3f}", end="")
    print()

print("\nTF-IDF feature extraction completed successfully.")
```
