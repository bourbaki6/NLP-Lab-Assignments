```python
import nltk
```

```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()
```

```python
words = [
    "playing", "played", "plays",
    "studying", "studied", "studies",
    "connected", "happiness"
]

print("\n Original Word -> Stemmed Word")

for word in words:
    print(word, " ->", stemmer.stem(word))
```
