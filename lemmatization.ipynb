```python
import nltk
```

```python
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()

words = [
    ("cats", "n"),
    ("students", "n"),
    ("studying", "v"),
    ("better", "a")
]

print("Original word -> Lemma")

for word, pos in words:
    print(word, "->", lemmatizer.lemmatize(word, pos = pos))
```
