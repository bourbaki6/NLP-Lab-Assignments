```python
import gensim.downloader as api
import numpy as np
```

```python
glove_model = api.load("glove-wiki-gigaword-100")
```

```python
word = "computer"
```

```python
if word in glove_model:
    vector = glove_model[word]

    print("\n Word: ", word)
    print("\n Embedding Vector:")
    print(vector)

    print("\n Vector Dimensions: ", len(vector))

else:
    print("\n Word not found in Vocab")
```

```python
word = "computer"

print("\n Words most similar to:", word)

for similar_word, similarity in glove_model.most_similar(word, topn = 5):
    print(similar_word, "- >", round(similarity, 4))

word1 = "king"
word2 = "queen"

similarity = glove_model.similarity(word1, word2)

print("\n Similarity between",word1, "word2", ":")
print(round(float(similarity), 4))

print("\n GloVe implemented.")
```
